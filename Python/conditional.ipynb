# if-elif-else in Python
**Definition**

if-elif-else is used to check multiple conditions.

if → checks first condition
elif → checks another condition
else → runs if all conditions are false

```python
# Syntax
if condition1:
    # code

elif condition2:
    # code

elif condition3:
    # code

else:
    # code
```

# while loop in python
**Definition:**
A while loop is used to run a block of code repeatedly until the condition becomes False.

```python
num = 1

while num <= 10:
    print(num)

    if num == 10:
        break

    num += 1
```

```python
num = 1

while num <= 10:
    print(num)

    if num == 10:
        continue

    num += 1
```

```python
num = 3
while num <=30:
    continue
    print(num)
    num +=3
```

```python
# number reverse
num = 5
while num >=1:
    print(num)
    num -=1
```

```python
num = 1

while num <= 10:
    print(num*2)

    num += 1
```

```python
num = 1
sum = 0
while num <= 5:
    sum+=num

    # print(num)
    num +=1
print(sum)

```

```python
num = 1
mul = 1

while num <= 5:
    mul *= num
    num += 1

print(mul)
```

```python
num = 7
fact = 1
while num >=1:
    fact= fact*num
    num -=1
print("factorial",fact)
```

# Uses of `while` Loop

## English

* Used when the number of iterations is not known.
* Runs until the condition becomes false.
* Used in menu programs and game loops.

* Jab repeat kitni baar hoga pata na ho tab use hota hai.
* Condition false hone tak loop chalta hai.
