# Uses of Class in Python
1. Create Objects

Class is used to create objects.
(Class ka use objects create karne ke liye hota hai).

2. Organize Code

It keeps data and functions in one place.
(Data aur functions ko ek jagah rakhne me help karta hai.)

3. Reuse Code

Same class can be used many times.
(Ek hi class ko baar-baar use kar sakte hain.)

4. Makes Program Easy

Programs become simple and easy to manage.
(Program simple aur manage karna easy ho jata hai.)

5. Provides Security

Using OOP concepts, data can be protected.
(OOP concepts ki help se data ko secure rakh sakte hain.)
