```python


def factorial(num):
    fact = 1
    
    while num >= 1:
        fact = fact * num
        num -= 1
    
    print("Factorial =", fact)

factorial(5)
```

# Return
In Python, return keyword function se value wapas bhejne ke liye use hota hai.
Jab function ka kaam complete ho jata hai, tab return result ko function call wali jagah par bhej deta hai.

# Uses of Return
* return function se value wapas bhejne ke liye use hota hai.
* Ye function ka result deta hai.
* return ke baad function stop ho jata hai.

```python
def add():
    a =10
    b = 10
    return a*b
print(add())

#Yahaan:

# add() → function ka naam hai
# return a * b → actual operation hai

# Isliye output 100 aaya, kyunki * multiplication karta hai.
```

```python
def mango():
    return 5 %2 # any operation are apply(👉👉👉+,-,*,%,/,//,**,^👉👉👉👉) 


print(mango())
```

```python
def find_min(lst):
    print(min(lst))
find_min([1,2,3,15,16])
```

```python
def find_sum(lst):
    print(sum(lst))
find_sum([1,2,3,4,5])
```

```python
def find_len(lst):
    print(len(lst))
find_len([1,2,3,4,5,6,7,8,98])
```
