# Encapsulation
Encapsulation is an Object-Oriented Programming (OOP) concept in which data and functions are wrapped together into a single unit called a class. It also helps to hide and protect data from direct access.


| Type of Encapsulation | Symbol | Access Level | Example |
|----------------------|--------|--------------|---------|
| Public               | No underscore | Accessible from anywhere | `name = "Saloni"` |
| Protected            | `_` Single underscore | Accessible inside class and child class | `_marks = 80` |
| Private              | `__` Double underscore | Accessible only inside the class | `__marks = 90` |

```python
class College:
    college_library="yes" #### Public
    account_information="2435657896543"#### Public
    College_name="CIT"#### Public

cr=College()
cr.college_library
cr.account_information
cr.College_name
```

```python
class College:
    college_library="yes" #### Public
    _account_information="2435657896543"#### private
    College_name="CIT"#### Public

cr=College()
cr.college_library
cr.account_information
cr.College_name
```

```python
class College:
    __college_library="yes" #### Protected
    _account_information="2435657896543"#### Private
    College_name="CIT"#### Public

cr=College()
cr.college_library
# cr.account_information
cr.College_name
```
