```python
from abc import ABC, abstractmethod
class School(ABC):
    @abstractmethod
    def display(self):
        print("hello")
    def show(self):
        print("hey")
class Teacher(School):
    def display(self):
        print("hello")
    def show(self):
        print("hey")


# s=School()
tch=Teacher()
tch.display()
tch.show()
```
