# Inheritance
A class can use the properties and methods of another class.
The class that gives features is called the Parent class.
The class that takes features is called the Child class.

👉👉👉 Ek class dusri class ki properties aur functions use kar sakti hai.👉👉👉Child class, Parent class se cheezein “inherit” karti hai.

# Singal inheritance
**Definition**
Single inheritance is a type of inheritance where one child class inherits from one parent class.

```python
class Father:
    pass
class Son(Father):
    pass
```

```python
class Father:
    father_name="santosh"
    father_account="1323454645"
class Son(Father):
    son_name="saloni"
beta =Son()
print(beta.father_name)
print(beta.father_account)
print(beta.son_name)
```

# Multilevel Inheritance
👉 One class inherits another class, and then a third class inherits the second class.

**Grandparent → Parent → Child**

       Grandparent
            ↓
         Parent
            ↓
           Child

```python
class Grandfather:
    pass
class Father:
    pass
class Son:
    pass
```

```python
class GrandFather:
    grand_father_balance="24354645756"
class Father(GrandFather):
    father_balance="215423534646"
class Son(Father):
    print("mai aapne papa ke paise per mojj kerungi")

beta = Son()
beta.grand_father_balance
beta.father_balance
```

```python
class Grandfather:
    def house(self):
        print("Grandfather has a house")

class Father(Grandfather):
    def car(self):
        print("Father has a car")

class Son(Father):
    def bike(self):
        print("Son has a bike")

obj = Son()

obj.house()
obj.car()
obj.bike()
```

```python
class Employee:
    def company_info(self):
        print("company name: TechInfo")

#child class
class Developer(Employee):
    def developer_work(self):
        print("developer writes and text code")

# grand child class
class TeamLead(Developer):
    def manage_team(self):
        print("team lead manages the developement team")

# object Creation
obj = TeamLead()
obj.company_info()
obj.developer_work()
obj.manage_team()
```

# Multiple Inheritance
👉 One child class inherits from more than one parent class.

        Father          Mother
           \              /
            \            /
             \          /
               Child

```python
class Father:
    pass
class Uncle:
    pass
class Son(Father,Uncle):
    pass
```

```python
class Father:
    father_balance=244324252
class Uncle:
    uncle_balance=4364765786
class Son(Father,Uncle):
    son_balance=0
st =Son()
print("father ke paise",st.father_balance)
print("uncle ke paise",st.uncle_balance)
print("son ke paise",st.son_balance)
```

```python
class Employee:
    def company_info(self):
        print("company name: TechInfo")

#child class
class Developer:
    def developer_work(self):
        print("developer writes and text code")

# grand child class
class TeamLead:
    def manage_team(self):
        print("team lead manages the developement team")



class Founder(Employee,Developer,TeamLead):
    def find_the_mistake(self):
        print("no problem")       

# object Creation
obj = Founder()

obj.company_info()
obj.developer_work()
obj.manage_team()
obj.find_the_mistake()
```

# Hybrid Inheritance
**Definition**
Hybrid inheritance is a combination of two or more types of inheritance.
It combines:

1. Multiple inheritance
2. Multilevel inheritance
3. Hierarchical inheritance



```python
class A:
    def showA(self):
        print("this is class A")

class B(A):
    def showB(self):
        print("this is my class B")
class C:
    def showC(self):
        print("this is my class C")
class D(B,C):
    def showD(self):
        print("this is my class D")
obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()


```

```python
class Employee:
    def company_info(self):
        print("company name: TechInfo")

#child class
class Developer:
    def developer_work(self):
        print("developer writes and text code")

# grand child class
class TeamLead:
    def manage_team(self):
        print("team lead manages the developement team")



class Founder(Employee,Developer):
    def find_the_mistake(self):
        print("no problem")       

# object Creation
obj = Founder()

obj.company_info()
obj.developer_work()
# obj.manage_team()
obj.find_the_mistake()
```

# Hierarchical Inheritance
Definition

Hierarchical inheritance is a type of inheritance where:

👉 Multiple child classes inherit from one parent class.

                Parent
               /      \
              /        \
            Child1    Child2

```python
class Papa:
    def show_papa(self):
        print("this is my papa class")

class Child1(Papa):
    def show_child1(self):
        print("this is my child1 class")

class Child2(Papa):
    def show_child2(self):
        print("this is my child2 class")


obj1 = Child1()
obj2 = Child2()

obj1.show_papa()
obj1.show_child1()

obj2.show_papa()
obj2.show_child2()
```

```python
#Heirachical inheritance
class papa:
    papa_name="Baap"
    sur_name="Pandey"
class Son1(papa):
    son_name="saloni"
class Son2(papa):
    son2_name="prince"

s1=Son1()
s2=Son2()

print(f"My name is {s1.son_name} {s1.sur_name}")
print(f"My name is {s2.son2_name} {s2.sur_name}")
```


# 👍👍Uses of Inheritance with Explanation

1. Code Reusability

Same code can be used again in different classes.

(Ek hi code ko baar-baar use kar sakte hain.)

2. Saves Time

No need to write the same code repeatedly.

(Same code dubara likhne ki zarurat nahi padti.)

3. Reduces Duplicate Code

Inheritance avoids writing duplicate functions and methods.

(epeated code likhne se bach sakte hain)

4. Makes Program Easy

Programs become simple, clean, and easy to understand.

(Code simple aur samajhne me easy ho jata hai.)

5. Adds New Features Easily

Child class can use parent features and also add new features.

(Child class parent ke features use karke naye features add kar sakti hai.)

6. Improves Code Organization

Code becomes properly structured and easier to manage.

(Program ka structure clean aur manageable rehta hai.)
