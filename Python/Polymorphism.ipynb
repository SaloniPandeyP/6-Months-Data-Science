# Polymorphism in Python

# Types of function polymorphism

```python
# Function Poly
print(len("hell"))
print(len([1,23,34,45]))
```

```python
# Method poly
class College:
    def show(self):
        print("hello")
    def show(self):
        print("hey")
College().show()
```

```python
class College:
    def show(self):
        print("hello")
class Student:
    def show(self):
        print("hey")
College().show()
Student().show()
```

```python
# Operator poly
print(10 + 20)          # Addition
print("Hello " + "World")  # String joining
print([1, 2] + [3, 4])       # List merging
```

```python
#Method overriding
class Company:
    def details(self):
        print("sab kam ker rhe h ")
class Developer:
    def details(self):
        print("all working")
Developer().details()
```
