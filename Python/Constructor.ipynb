# Constructor in Python

A constructor is a special method that automatically runs when an object of a class is created.

In Python, constructor is written as  __init__ ().

````python
#Synatx 
class School:
    def __init__(self):
        pass
````

````python
#Synatx 
class School:
    def __init__(self):
        pass

student=School()
````

````python
class School:
    def __init__(self):
        self.student_name="saloni"### instance variable
        self.classes ="fourth year"### instance variable
        self.branch ="CSE"### instance variable
        self.rollnumber=29### instance variable

student=School()
student1=School()
student2=School()


print(student.branch)
print(student.classes)
print(student.rollnumber)
print(student.student_name)

````

| Feature | Instance Variable | Class Variable |
|---|---|---|
| Definition | Variable created inside constructor using `self` | Variable created directly inside class |
| Belongs To | Object | Class |
| Separate Copy | Yes, each object has different copy | No, all objects share same copy |
| Access | Using object | Using class or object |
| Example | `self.name = "Saloni"` | `college = "ABC"` |

````python
#task multiple const create

class Student:

    # Main Constructor
    def __init__(self, name=None, age=None):

        # No argument
        if name is None and age is None:
            print("Default Constructor")
            self.name = "Unknown"
            self.age = 0

        # One argument
        elif age is None:
            print("One Parameter Constructor")
            self.name = name
            self.age = 0

        # Two arguments
        else:
            print("Two Parameter Constructor")
            self.name = name
            self.age = age

    def show(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print()


# Object 1
s1 = Student()
s1.show()

# Object 2
s2 = Student("Rahul")
s2.show()

# Object 3
s3 = Student("Rahul", 20)
s3.show()
````

# super() Keyword in Python

super() keyword ka use parent class ke methods ya constructor ko child class ke andar access karne ke liye hota hai.

| Use                     | Explanation                                 |
| ----------------------- | ------------------------------------------- |
| Parent constructor call | Parent class ka constructor chalane ke liye |
| Parent method access    | Parent methods use karne ke liye            |
| Code reusability        | Same code dubara likhne ki zarurat nahi     |
| Inheritance support     | Parent-child relation manage karta hai      |


````python
# Parent class
class Animal:
    def sound(self):
        print("Animals make sounds")

# Child class
class Dog(Animal):
    def sound(self):
        super().sound()   # Calls parent method
        print("Dog barks")

# Object creation
d = Dog()
d.sound()
````

````python
 # Parent class
class Employee:
    def work(self):
        print("Employee works in the company")

# Child class
class Developer(Employee):
    def work(self):
        super().work()   # Parent class method call
        print("Developer writes code")

# Object creation
d = Developer()
d.work()
````
