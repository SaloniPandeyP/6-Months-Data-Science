```python
try:
    num1=int(input("enter your first num1"))
    num2=int(input("enter your second num2"))
    print(num1/num2)
except ZeroDivisionError as e:
    print("cannot divide by zero")
except ValueError:
    print("plz enter vaild number")
```

```python
try:
    a = 10
    b = "eww"
    print(a+b)
except TypeError as e:
    print(e)
```

```python
try:
    lst = [1,23,4,5]
    print(lst[10])
except IndexError as t:
    print(t)
```

```python
try:
    a = 10
    b = "eww"
    print(a+b)
except Exception as e:
    print(e)
```
